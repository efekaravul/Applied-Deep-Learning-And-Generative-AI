# PyTorch'ta Cihaz Ayarları / Device Settings in PyTorch

🇹🇷 PyTorch tensorleri CPU ya da GPU üzerinde tutulabilir. Bu defterde GPU'nun kullanılabilir olup
olmadığını nasıl kontrol edeceğimizi, cihazdan bağımsız (device-agnostic) kod yazmayı ve
tensorleri cihazlar arasında taşımayı görüyoruz.

🇬🇧 PyTorch tensors can live on the CPU or on the GPU. In this notebook we see how to check whether
a GPU is available, how to write device-agnostic code, and how to move tensors between devices.

```python
import torch
```

## GPU var mı? / Is a GPU available?

🇹🇷 `torch.cuda.is_available()` sistemde CUDA destekli bir GPU olup olmadığını söyler. Sonuç
`False` ise kod CPU üzerinde çalışmaya devam eder, sadece daha yavaş olur.

🇬🇧 `torch.cuda.is_available()` tells us whether a CUDA-capable GPU is present. If it returns
`False` the code still runs on the CPU, just more slowly.

```python
torch.cuda.is_available()
```

## Cihazdan bağımsız kod / Device-agnostic code

🇹🇷 Cihazı tek bir değişkende toplamak, aynı defterin hem GPU'lu hem GPU'suz makinelerde
değişiklik yapmadan çalışmasını sağlar. Bu kalıp kurs boyunca tekrar tekrar kullanılacak.

🇬🇧 Collecting the device in a single variable lets the same notebook run unchanged on machines
with and without a GPU. This pattern will be used again and again throughout the course.

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
device
```

## Tensorü cihaza taşımak / Moving a tensor to a device

🇹🇷 Tensorler varsayılan olarak CPU'da oluşturulur; `.device` bunu gösterir. `.to(device)` tensorün
hedef cihazdaki bir kopyasını döndürür, orijinali yerinde bırakır.

🇬🇧 Tensors are created on the CPU by default, as `.device` shows. `.to(device)` returns a copy of
the tensor on the target device and leaves the original where it was.

```python
tensor = torch.tensor([1,2,3])
```

```python
tensor.device
```

```python
tensor_on_gou = tensor.to(device)
```

## GPU'dan CPU'ya geri dönmek / Coming back from GPU to CPU

🇹🇷 NumPy yalnızca CPU belleğiyle çalışır; bu yüzden GPU'daki bir tensoru NumPy'a çevirmeden ya da
çizdirmeden önce `.cpu()` ile geri taşımak gerekir.

🇬🇧 NumPy only works with CPU memory, so a tensor on the GPU has to be moved back with `.cpu()`
before converting it to NumPy or plotting it.

```python
tensor_on_gou.cpu()
```

## Cihaz context manager'ı / The device context manager

🇹🇷 `with torch.device(device):` bloğu içinde oluşturulan tüm tensorler ve katmanlar doğrudan o
cihazda üretilir. Böylece her nesne için ayrı ayrı `.to(device)` yazmak gerekmez.

🇬🇧 Every tensor and layer created inside a `with torch.device(device):` block is placed on that
device directly, so we do not have to write `.to(device)` for each object separately.

```python
with torch.device(device):
    tensor2 = torch.tensor([1,2,3])
    layer = torch.nn.Linear(20,30)
```

```python
tensor2.device
```
