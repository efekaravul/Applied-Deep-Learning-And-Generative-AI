# Rastgele Tensorler / Random Tensors

🇹🇷 Sinir ağlarının ağırlıkları rastgele değerlerle başlatılır, bu yüzden rastgele tensor üretmek
temel bir işlemdir. Bu defterde NumPy ve PyTorch'un rastgele üretim fonksiyonlarını karşılaştırıyoruz.

🇬🇧 Neural network weights are initialised with random values, which makes generating random
tensors a fundamental operation. In this notebook we compare the random generators of NumPy and
PyTorch.

```python
import torch
import numpy as np
```

## NumPy ile rastgele dizi / Random array with NumPy

🇹🇷 `np.random.randn(3,5)` standart normal dağılımdan 3x5'lik bir dizi üretir. Sonuç bir
`numpy.ndarray`'dir; boyutu `ndim`, şekli `shape` ile okunur.

🇬🇧 `np.random.randn(3,5)` produces a 3x5 array drawn from the standard normal distribution. The
result is a `numpy.ndarray`; its dimensionality is read with `ndim` and its shape with `shape`.

```python
random_array = np.random.randn(3,5)
```

```python
random_array
```

```python
type(random_array)
```

```python
random_array.shape
```

```python
random_array.ndim
```

## PyTorch ile rastgele tensor / Random tensor with PyTorch

🇹🇷 `torch.rand(size=(3,5))` [0,1) aralığında düzgün (uniform) dağılımlı bir tensor üretir.
NumPy'deki `randn`'den farkı: `randn` normal dağılım, `rand` düzgün dağılım kullanır.

🇬🇧 `torch.rand(size=(3,5))` produces a tensor with a uniform distribution over [0,1). The
difference from NumPy's `randn`: `randn` uses a normal distribution, `rand` a uniform one.

```python
random_matrix = torch.rand(size=(3,5))
```

```python
random_matrix
```

```python
random_matrix.ndim
```

```python
random_matrix.shape
```

```python
random_matrix.dtype
```

## Görüntü şeklinde rastgele tensor / Random image-shaped tensor

🇹🇷 `(224, 224, 3)` şekli tipik bir görüntüyü temsil eder: 224 piksel yükseklik, 224 piksel
genişlik ve 3 renk kanalı (RGB). Görüntü modelleri verilerini bu şekilde tensor olarak alır.

🇬🇧 The shape `(224, 224, 3)` represents a typical image: 224 pixels high, 224 pixels wide and 3
colour channels (RGB). Vision models take their input as tensors in exactly this form.

```python
random_image_tensor = torch.rand(size=(224,224,3))
```

```python
random_image_tensor
```

```python
random_image_tensor.shape
```

```python
random_image_tensor.ndim
```
