```python
import torch
import numpy as np
```

## Sıfırlar ve birler / Zeros and ones

🇹🇷 `torch.zeros` ve `torch.ones` verilen şekilde tamamı 0 veya 1 olan tensorler üretir. Bir
tensoru başlangıç değeriyle doldurmak ya da maske oluşturmak için kullanılır.

🇬🇧 `torch.zeros` and `torch.ones` create tensors of the given shape filled entirely with 0 or 1.
They are used to initialise a tensor with a starting value or to build masks.

```python
zeros = torch.zeros(size = (3,2))
```

```python
zeros
```

```python
ones = torch.ones(size = (3,2))
ones
```

## Aralık üretme / Creating a range

🇹🇷 `torch.arange(start, end, step)` belirtilen adımla artan bir vektör üretir. `end` değeri
aralığa dahil değildir.

🇬🇧 `torch.arange(start, end, step)` produces a vector increasing by the given step. The `end`
value is exclusive.

```python
vector_with_arange = torch.arange(start=0,end=10,step=2)
```

```python
vector_with_arange
```

## Başka bir tensorün şeklini kopyalama / Copying another tensor's shape

🇹🇷 `torch.ones_like(input=...)` verilen tensorle aynı şekle sahip, tamamı 1 olan yeni bir tensor
üretir. Şekli elle yazmadan aynı boyutta tensor oluşturmanın kısa yoludur.

🇬🇧 `torch.ones_like(input=...)` creates a new tensor of ones with the same shape as the given
tensor. It is a shortcut for matching a shape without writing it out by hand.

```python
similar_ones = torch.ones_like(input=vector_with_arange)
```

```python
similar_ones
```

## Skaler ile işlemler (yayınlama) / Operations with a scalar (broadcasting)

🇹🇷 Bir tensor ile tek bir sayı işleme sokulduğunda, sayı tüm elemanlara ayrı ayrı uygulanır.
Buna yayınlama (broadcasting) denir. Ondalıklı bir sayıyla çarpmak sonucun tipini float yapar.

🇬🇧 When a tensor is combined with a single number, the number is applied to every element
separately. This is called broadcasting. Multiplying by a decimal number turns the result into a
float tensor.

```python
x_tensor = torch.tensor([1,2,3])
```

```python
x_tensor + 15
```

```python
x_tensor * 3
```

```python
x_tensor * 3.14
```

## Operatörler ve metot karşılıkları / Operators and their method forms

🇹🇷 `*`, `+`, `/` operatörlerinin metot karşılıkları vardır: `multiply()`, `add()`, `divide()`.
İkisi de aynı işi yapar; metot biçimi zincirleme çağrılarda daha okunaklı olabilir.

🇬🇧 The `*`, `+` and `/` operators have method equivalents: `multiply()`, `add()`, `divide()`.
Both do the same job; the method form can read better in chained calls.

```python
x_tensor.multiply(10)
```

```python
x_tensor.add(20)
```

```python
x_tensor.divide(2)
```

## İki tensor arasında eleman bazlı işlemler / Element-wise operations between two tensors

🇹🇷 Aynı şekle sahip iki tensor arasındaki işlemler eleman eleman yapılır: birinci elemanla
birinci, ikinciyle ikinci... Bu, matris çarpımından farklıdır.

🇬🇧 Operations between two tensors of the same shape are applied element by element: first with
first, second with second, and so on. This is different from matrix multiplication.

```python
x_tensor * x_tensor
```

```python
x_tensor + x_tensor
```

```python
x_tensor + x_tensor
```

```python
x_tensor / x_tensor
```

iki vektör (ör. 3 ve 2 elemanlı) doğrudan işleme sokulamaz.

🇬🇧 Element-wise operations require compatible shapes. Two vectors of different lengths (e.g. one
with 3 and one with 2 elements) cannot be combined directly.
